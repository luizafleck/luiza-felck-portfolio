```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Загрузка данных
file_path = '//Users//wejey//Загрузки//Life Expectancy Data.csv'  
df = pd.read_csv(file_path)
```

```python
# Выводим все названия столбцов, чтобы увидеть их точные имена
print("Названия столбцов:")
print(df.columns)
```

```python
# Общая информация о датасете
print("\nОбщая информация о датасете:")
print(df.info())
```

```python
# Проверка пропущенных значений
print("\nПропущенные значения:")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])
```

```python
# Описание числовых данных
print("\nСтатистика по числовым данным:")
df.describe()
```

```python
# Убедимся, что столбец "Life expectancy" существует
if 'Life expectancy ' in df.columns:
    # Заполнение пропущенных значений медианой (с использованием numpy)
    df['Life expectancy '] = df['Life expectancy '].fillna(np.median(df['Life expectancy '].dropna()))
else:
    print("Столбец 'Life expectancy' не найден в данных!")

if 'Adult Mortality' in df.columns:
    df['Adult Mortality'] = df['Adult Mortality'].fillna(np.median(df['Adult Mortality'].dropna()))
else:
    print("Столбец 'Adult Mortality' не найден в данных!")
```

```python
# Корреляция между числовыми переменными
corr_matrix = df.corr(numeric_only=True)
```

```python
# Построение тепловой карты для визуализации корреляций
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title('Корреляционная матрица')
plt.show()
```

```python
# Дополнительные визуализации: гистограмма по продолжительности жизни
plt.figure(figsize=(10, 6))
sns.histplot(df['Life expectancy '], kde=True, color='skyblue', bins=20)
plt.title('Распределение продолжительности жизни')
plt.xlabel('Ожидаемая продолжительность жизни')
plt.ylabel('Частота')
plt.show()
```

```python
# Корреляция продолжительности жизни с другими переменными
life_expectancy_corr = corr_matrix['Life expectancy '].sort_values(ascending=False)

print("\nКорреляция продолжительности жизни с другими переменными:")
print(life_expectancy_corr)

# Визуализация наиболее сильно коррелирующих переменных с продолжительностью жизни
top_corr_features = life_expectancy_corr[1:].head(10)  # Исключаем саму продолжительность жизни
plt.figure(figsize=(10, 6))
top_corr_features.plot(kind='bar', color='lightgreen')
plt.title('Топ-10 факторов, наиболее сильно коррелирующих с продолжительностью жизни')
plt.xlabel('Переменные')
plt.ylabel('Корреляция')
plt.xticks(rotation=45)
plt.show()
```
